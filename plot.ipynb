```python
import matplotlib.pyplot as plt
import textwrap

questions = [
    "How clear is the whole idea of integrating inventive problem-solving techniques into software development life cycles with TRIZ?",
    "How useful is the TRIZ AI tool in guiding you through the problem-solving process?",
    "How clear and easy to follow were the steps from the problem-solving process?",
    "Did you understand the logic of the workflow?",
    "Is it clear what the Ideal Final Result (IFR) means?",
    "Is it clear what the Technical Contradictions (TCs) mean?",
    "How satisfied were you with the Ideal Final Result generated for your problem?",
    "Did you find the comparison between the Ideal Final Result and the State-of-the-Art useful?",
    "Did you feel that taking one step back from the Ideal Final Result helped in identifying any missed opportunities or improvements?",
    "Were the Technical Contradictions generated by the tool relevant and helpful?",
    "Do you think that the explanations and guidance provided by the tool were enough in understanding and implementing TRIZ principles?",
    "Are you satisfied with the solutions generated by the TRIZ AI tool?",
    "How helpful was the comparison between your selected solution and the State-of-the-Art?"
]

values = [7.15, 7.45, 7.45, 7.5, 8.5,
          7.2, 7.6, 7.15, 6.55,
          7.25, 7, 7.1, 7.25]

plt.figure(figsize=(12, 8))
bars = plt.barh(range(len(questions)), values, color='skyblue')
plt.xlabel('The Average Value of User Satisfaction with Features')
plt.ylabel('Survey Questions')
plt.title('Bar Chart of Numerical Survey Responses')
plt.gca().invert_yaxis()  

plt.xlim(0, 9)

plt.yticks(range(len(questions)), [textwrap.fill(question, 50) for question in questions])

for bar, value in zip(bars, values):
    plt.text(bar.get_width(), bar.get_y() + bar.get_height()/2, f'{value:.2f}', 
             va='center', ha='left', fontsize=8)

plt.show()
```
